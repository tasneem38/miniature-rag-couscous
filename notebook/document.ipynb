```python
sample_texts={
"../data/text_files/rag_intro.txt":"""RAG (Retrieval-Augmented Generation) Overview
===============================================
Introduction
------------
Retrieval-Augmented Generation (RAG) is a technique that combines
information retrieval with text generation. It enhances large language
models (LLMs) by providing them with relevant external knowledge
retrieved from a knowledge base or document store.

Key Components
--------------
1. Document Loader - Loads documents from various sources (PDF, TXT, etc.)
2. Text Splitter - Splits documents into manageable chunks
3. Embeddings Model - Converts text chunks into vector embeddings
4. Vector Store - Stores embeddings for efficient similarity search
5. Retriever - Retrieves relevant chunks based on user queries
6. LLM - Generates responses using retrieved context as source material

How RAG Works
-------------
1. A user submits a query
2. The query is converted into an embedding vector
3. The vector store performs similarity search to find relevant chunks
4. Retrieved chunks are passed as context to the LLM
5. The LLM generates a response grounded in the retrieved information

Benefits
--------
- Reduces hallucinations by grounding responses in actual data
- Enables domain-specific question answering
- No need to retrain the model for new information
- Cost-effective compared to fine-tuning

Use Cases
---------
- Customer support chatbots
- Enterprise knowledge base Q&A
- Academic research assistants
- Legal document analysis
- Medical information retrieval""",

"../data/text_files/machine_learning.txt":"""Machine Learning (ML) is a branch of Artificial Intelligence (AI) that enables computers to learn patterns from data and make predictions or decisions without being explicitly programmed for every task.

It works by training a model on historical data so it can recognize patterns and apply them to new, unseen data.

Common types of Machine Learning:

Supervised Learning: Learns from labeled data (e.g., spam email detection).
Unsupervised Learning: Finds hidden patterns in unlabeled data (e.g., customer segmentation).
Reinforcement Learning: Learns through trial and error using rewards and penalties (e.g., game-playing AI).

Applications:

Recommendation systems (Netflix, Amazon)
Fraud detection
Image and speech recognition
Predictive maintenance
Medical diagnosis

In simple terms, machine learning helps computers improve their performance by learning from data instead of following fixed rules."""
}
for filepath,content in sample_texts.items():
    with open(filepath,"w",encoding="utf-8") as f:
        f.write(content)
print("end=> created sample text files in data/text_files")
```

```python
### text loader
from langchain_community.document_loaders import TextLoader
loader = TextLoader("../data/text_files/rag_intro.txt", encoding="utf-8")
loader.load()
```

```python
### directory loader
from langchain_community.document_loaders import DirectoryLoader
dir_loader = DirectoryLoader(
    "../data/text_files", 
    glob="**/*.txt", ##pattern to match files
    loader_cls=TextLoader, #loader class to use
    loader_kwargs={"encoding": "utf-8"},
    show_progress=False
)
documents = dir_loader.load()
print(documents)
```

```python
### pdf loader
from langchain_community.document_loaders import PyPDFLoader, PyMuPDFLoader 
dir_loader = DirectoryLoader(
    "../data/pdf", 
    glob="**/*.pdf", ##pattern to match files
    loader_cls=PyPDFLoader, #loader class to use
    show_progress=False
)
pdf_doc = dir_loader.load()
print(pdf_doc)
```
